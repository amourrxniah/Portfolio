```python
#import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from sklearn.preprocessing import LabelEncoder
import warnings
warnings.filterwarnings('ignore')

print("All libraries imported successfully!")
```

```python
#load the dataset
def load_mental_health_data():
    """Load and preprocess the mental health in tech data"""
    try:
        #load dataset
        df = pd.read_csv('data/responses.csv')
        print(f"Dataset loaded successfully with shape: {df.shape}")
        return df
    except FileNotFoundError:
        print("Error: Dataset file not found.")
        return None

#load data
df = load_mental_health_data()
```

```python
if df is not None:
    #preview first 5 rows
    print("First 5 rows of the dataset:")
    display(df.head())

    #basic info
    print("\nDataset Info:")
    print(f"Shape: {df.shape}")
    print(f"Columns: {len(df.columns)}")
    print("\nColumn names:")
    print(df.columns.tolist())

    #check for missing values
    print("\nMissing values summary:")
    missing_data = df.isnull().sum()
    print(missing_data[missing_data > 0])

    #select key columns for analysis
    key_columns = [
        'What is your age?',
        'What is your gender?',
        'How many employees does your company or organisation have?',
        'Is your employer primarily a tech company/organszation?',
        'Is your primary role within your company related to tech/IT?',
        'Does your employer provide mental health benefits as part of healthcare coverage?',
        'Have you ever discussed your mental health with your employer?',
        'Would you feel comfortable discussing a mental health issue with your coworkers?',
        'Do you have medical coverage (private insurance or state-provided) that includes treatment for mental health disorders?',
        'Do you believe your productivity is ever affected by a mental health issue?',
        'Have you ever been *diagnosed* with a mental health disorder?',
        'What country do you *live* in?'
    ]

    #create a subset with key columns
    analysis_df = df[key_columns].copy()

    #clean column names
    analysis_df.columns = [
        'age', 'gender', 'company_size', 'is_tech_company', 'is_tech_role',
        'mental_health_benefits', 'discussed_with_employer', 'comfortable_with_coworkers',
        'mental_health_coverage', 'productivity_affected', 'diagnosed', 'country'
    ]

    print("\nCleaned dataset for analysis:")
    display(analysis_df.head())

    #basic statistics
    print("\nDescriptive Statistics:")
    print(analysis_df.describe())

    print("\nCategorical Variables Summary:")
    categorical_cols = ['gender', 'company_size', 'is_tech_company', 'is_tech_role',
                        'mental_health_benefits', 'discussed_with_employer', 'comfortable_with_coworkers',
                        'mental_health_coverage', 'productivity_affected', 'diagnosed']

    for col in categorical_cols:
        print(f"\n{col}:")
        print(analysis_df[col].value_counts())
        print("-" * 50)
```

```python
class MentalHealthAnalyser:
    def __init__(self):
        self.df = self.generate_sample_data()

    def generate_sample_data(self):
        """Generate sample mental health dataset"""
        np.random.seed(42)
        n_samples = 1000

        data = {
            'age': np.random.randint(18, 65, size=n_samples),
            'gender': np.random.choice(['Male', 'Female', 'Other'], n_samples),
            'company_size': np.random.choice(['1-50', '51-200', '201-1000', '1000+'], n_samples),
            'tech_role': np.random.choice(['Developer', 'Data Scientist', 'Manager', 'Designer'], n_samples),
            'work_hours': np.random.randint(35, 80, n_samples),
            'remote_work': np.random.choice(['Yes', 'No'], n_samples),
            'mental_health_support': np.random.choice(['Yes', 'No', 'Not Sure'], n_samples),
            'stress_level': np.random.randint(1, 10, n_samples),
            'productivity_level': np.random.randint(1, 10, n_samples),
        }

        return pd.DataFrame(data)
```

```python
#initialise the analyser
analyser = MentalHealthAnalyser()

#preview first 5 rows
print("First 5 rows of the dataset:")
display(analyser.df.head())
```

```python
#get basic info about dataset
print("Dataset Shape:", analyser.df.shape)
print("\nDataset Info:")
analyser.df.info()
```

```python
#descriptive statistics
print("Descriptive Statistics:")
display(analyser.df.describe())

print("\nCategorical Variables Summary:")
categorical_cols = ['gender', 'company_size', 'tech_role', 'remote_work', 'mental_health_support']
for col in categorical_cols:
    print(f"\n{col}:")
    print(analyser.df[col].value_counts())
```

```python
#set style for better plots
plt.style.use('default')
sns.set_palette("husl")

#create demographic plots
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

#gender distribution
analyser.df['gender'].value_counts().plot(kind='bar', ax=axes[0,0], title='Gender Distribution', color='skyblue')
axes[0,0].set_ylabel('Count')

#age distribution
analyser.df['age'].plot(kind='hist', bins=20, ax=axes[0,1], title='Age Distribution', color='lightgreen')
axes[0,1].set_xlabel('Frequency')

#company size
analyser.df['company_size'].value_counts().plot(kind='pie', ax=axes[1,0], title='Company Size Distribution', autopct='%1.1f%%')

#tech role
analyser.df['tech_role'].value_counts().plot(kind='bar', ax=axes[1,1], title='Tech Role Distribution', color='orange')
axes[1,1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
#calculate correlations between numerical variables
numerical_cols = ['age', 'work_hours', 'stress_level', 'productivity_level']
correlation_matrix = analyser.df[numerical_cols].corr()

print("Correlation Matrix:")
display(correlation_matrix)

#heatmap
plt.figure(figsize=(8,6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, square=True)
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

```python
#stress level by different categories
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

#stress by gender
sns.boxplot(data=analyser.df, x='gender', y='stress_level', ax=axes[0,0])
axes[0,0].set_title('Stress Level by Gender')

#stress by company size
sns.boxplot(data=analyser.df, x='company_size', y='stress_level', ax=axes[0,1])
axes[0,1].set_title('Stress Level by Company Size')

#stress by tech role
sns.boxplot(data=analyser.df, x='tech_role', y='stress_level', ax=axes[1,0])
axes[1,0].set_title('Stress Level by Tech Role')

#stress by remote work
sns.boxplot(data=analyser.df, x='remote_work', y='stress_level', ax=axes[1,1])
axes[1,1].set_title('Stress Level by Remote Work')

plt.tight_layout()
plt.show()
```
